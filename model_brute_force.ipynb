```python
import numpy as np
import pandas as pd
from optimal_transport import *
```

```python
shapes = generate_instances(67)
shape_names = ["disk", "sphere", "torus", "klein"]
```

```python
#Iterate over shapes
total_score = 0
weight = {"disk":1, "sphere":2, "torus":3, "klein":5}
for name in shape_names:
    #Constants
    total_tokens = 1000
    m = shapes[name]
    d_index = []
    #Sort by shortest distance
    for i, pt in enumerate(m.endpoints):
        d = geodesic_distance(m, m.start, pt)
        d_index.append((d, i))
    d_index = sorted(d_index)

    #Greedily allocate the horse/truck/jet for small distances
    all_deliveries = []
    for pair in d_index:
        vehicle = ""
        if pair[0] < 50 * 1000:
            vehicle = "horse"
        elif pair[0] < 750 * 1000:
            vehicle = "truck"
        else:
            vehicle = "jet"
        #evaluate potential delivery method
        delivery = evaluate_delivery(m, pair[1], vehicle, ())
        if total_tokens - delivery.total_tokens > 0:
            total_tokens -= delivery.total_tokens
            all_deliveries.append({"endpoint":pair[1], "vehicle":vehicle, "checkpoints":()})
    score_obj = score_manifold(m, all_deliveries, 1000)
    total_score += score_obj["score"] * weight[name]
print("Total Score Minimizing Distance: ", total_score/11)    
```
